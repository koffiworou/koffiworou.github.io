```python
import requests

# From  this main page: https://climatereanalyzer.org/wx/todays-weather/?var_id=sstanom&ortho=1&wt=1


url = "https://climatereanalyzer.org/wx/todays-weather/maps/gfs_world-wt_sstanom_d1.png"

r = requests.get(url)
r.raise_for_status()

figdir = "../images/downloaded_figures/"
import os
if not os.path.exists(figdir):
    os.makedirs(figdir)
filename = "gfs_world-wt_sstanom_d1.png"
filepath = figdir + filename

with open(filepath, "wb") as f:
    f.write(r.content)
```
